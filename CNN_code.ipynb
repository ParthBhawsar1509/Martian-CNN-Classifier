```python
import os
import zipfile
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping


# Step 1: Download and extract the dataset
url = 'https://zenodo.org/records/4002935/files/hirise-map-proj-v3_2.zip'
dataset_path = 'hirise-map-proj-v3_2.zip'
if not os.path.exists(dataset_path):
    print("Downloading dataset...")
    response = requests.get(url)
    with open(dataset_path, 'wb') as f:
        f.write(response.content)
    print("Download complete.")
# Extract to specific directory
extract_dir = 'new/hirise-map-proj-v3_2'
if not os.path.exists(extract_dir):
    print("Extracting dataset...")
    os.makedirs(extract_dir, exist_ok=True)
    with zipfile.ZipFile(dataset_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print("Extraction complete.")
# Paths adjusted
images_dir = os.path.join(extract_dir, 'map-proj-v3_2')
labels_file = os.path.join(extract_dir, 'labels-map-proj-v3_2_train_val_test.txt')
classmap_file = os.path.join(extract_dir, 'landmarks_map-proj-v3_2_classmap.csv')
# Load class map without headers
class_map = pd.read_csv(classmap_file, header=None)
class_names = dict(zip(class_map[0], class_map[1]))
num_classes = len(class_names)
print("Class names:", class_names)
# Load labels and splits
df = pd.read_csv('labels-map-proj_v3_2_train_val_test.txt', sep=' ', names=['filename', 'class_id', 'set'])
df['label'] = df['class_id'].map(class_names)
df['path'] = df['filename'].apply(lambda x: os.path.join(images_dir, x))
# Display some input data (10 images)
print("Displaying 10 sample images...")
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
axes = axes.ravel()
# Sample one from each class
sample_df = df.groupby('class_id').apply(lambda x: x.sample(1), include_groups=False).reset_index(drop=True)
# Find majority class (assuming 'other' is majority)
majority_class = df['class_id'].value_counts().idxmax()
# Sample two more from majority class
extra_df = df[df['class_id'] == majority_class].sample(2)
# Combine
sample_df = pd.concat([sample_df, extra_df]).reset_index(drop=True)
for i in range(10):
    row = sample_df.iloc[i]
    img = load_img(row['path'])
    axes[i].imshow(img)
    axes[i].set_title(row['label'])
    axes[i].axis('off')
plt.tight_layout()
plt.show()
# Preprocessing: Images are 227x227, check channels
sample_img = img_to_array(load_img(df['path'].iloc[0]))
input_shape = sample_img.shape
print("Image shape:", input_shape)
channels = input_shape[2] if len(input_shape) == 3 else 1
# Split data
train_df = df[df['set'] == 'train']
val_df = df[df['set'] == 'val']
test_df = df[df['set'] == 'test']
print(f"Train: {len(train_df)}, Val: {len(val_df)}, Test: {len(test_df)}")
# Generators
train_datagen = ImageDataGenerator(rescale=1./255)
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)
def create_generator(df, datagen, batch_size=32, shuffle=True):
    return datagen.flow_from_dataframe(
        df,
        x_col='path',
        y_col='class_id',
        target_size=(227, 227),
        batch_size=batch_size,
        class_mode='raw',
        shuffle=shuffle
    )
batch_size = 32
train_gen = create_generator(train_df, train_datagen, batch_size, shuffle=True)
val_gen = create_generator(val_df, val_datagen, batch_size, shuffle=False)
test_gen = create_generator(test_df, test_datagen, batch_size=1, shuffle=False)


# Step 2: Design CNN
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(227,227,channels)),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Conv2D(128, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
])
model.summary()
# Compile
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
# Early stopping
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
# Train
epochs = 50
history = model.fit(
    train_gen,
    epochs=epochs,
    validation_data=val_gen,
    callbacks=[early_stop]
)


# Step 3: Plots for training performance
plt.figure(figsize=(8,4))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Accuracy vs Epochs')
plt.show()
plt.figure(figsize=(8,4))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.title('Loss vs Epochs')
plt.show()


# Step 4: Evaluate on test set
test_loss, test_acc = model.evaluate(test_gen)
print(f"Test Accuracy: {test_acc:.2f}")
# Predictions
test_gen.reset()
preds = model.predict(test_gen, steps=len(test_df))
pred_classes = np.argmax(preds, axis=1)
true_classes = test_df['class_id'].values
# Classification report
print(classification_report(true_classes, pred_classes, target_names=list(class_names.values())))
# Accuracy per class
cm = confusion_matrix(true_classes, pred_classes)
per_class_acc = cm.diagonal() / cm.sum(axis=1)
plt.figure(figsize=(10,6))
sns.barplot(x=list(class_names.values()), y=per_class_acc)
plt.xticks(rotation=45)
plt.ylabel('Accuracy')
plt.title('Accuracy per Class')
plt.show()


# Step 5: Show 20 test images with true and pred labels, at least one per class
selected = test_df.groupby('class_id').apply(lambda x: x.sample(min(3, len(x))), include_groups=False).reset_index(drop=True)
if len(selected) < 20:
    extra = test_df.sample(20 - len(selected))
    selected = pd.concat([selected, extra]).drop_duplicates()
selected_imgs = np.array([img_to_array(load_img(p, target_size=(227,227))) for p in selected['path']])
selected_imgs /= 255.0
selected_preds = model.predict(selected_imgs)
selected_pred_classes = np.argmax(selected_preds, axis=1)
fig, axes = plt.subplots(4, 5, figsize=(20, 16))
axes = axes.ravel()
for i in range(20):
    if i < len(selected):
        img = load_img(selected['path'].iloc[i])
        axes[i].imshow(img)
        true_label = selected['label'].iloc[i]
        pred_label = class_names[selected_pred_classes[i]]
        axes[i].set_title(f"True: {true_label}\nPred: {pred_label}")
        axes[i].axis('off')
plt.tight_layout()
plt.show()
# Conclusions
print("The network performed with test accuracy of {:.2f}".format(test_acc))
```
